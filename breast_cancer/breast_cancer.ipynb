## 1. What Is Breast Cancer?

Breast cancer is a type of cancer that starts in the cells of the breast.

Normally, cells in our body follow a controlled process:

- They grow when needed.
- They divide to create new cells.
- They perform their specific functions.
- They stop dividing when appropriate.
- Damaged or old cells can be eliminated.

Breast cancer develops when cells in the breast begin to grow and divide in an abnormal and uncontrolled way.

These abnormal cells can accumulate and form a mass called a **tumor**.

Breast cancer can begin in different tissues of the breast. Many breast cancers begin in the cells lining the **milk ducts**, while others begin in the **lobules**, the glands that produce milk.

<h3 style="color:#8E44AD">Simple idea</h3>

```
Normal breast cell
↓
Genetic / cellular changes
↓
Abnormal growth and division
↓
Accumulation of abnormal cells
↓
Tumor
↓
Breast cancer
```
### Important

Breast cancer is not usually explained by one single cause.

Scientists identify **risk factors** that can increase the probability of developing breast cancer, but having a risk factor does not mean that a person will definitely develop cancer.

## 2. How Does Breast Cancer Develop?

To understand cancer, we first need to understand that a normal cell has several control mechanisms.

A healthy cell must know:

- when to grow
- when to divide
- when to stop dividing
- how to repair DNA damage
- when to die if it is severely damaged

These controls keep the number and behavior of cells balanced.

Cancer therefore involves changes that disrupt the normal control of cell growth and division.

## 3. What Is DNA?

### DNA = Deoxyribonucleic Acid

DNA is the molecule that stores genetic information in our cells.

### Think of DNA like an instruction system

- **Genes** → specific sections of those instructions
- **Proteins** → molecules produced using genetic information
- **Cell** → the biological system that uses these instructions

Genes are sections of DNA that contain information involved in producing proteins or certain RNAs and in controlling cellular functions.

Some genes are involved in important processes such as:

- cell growth
- cell division
- DNA repair
- cell survival and death

This is why changes to DNA can sometimes affect how a cell behaves.

## 4. What Does a DNA Change Mean?

DNA is built from four chemical bases:

- A = Adenine
- T = Thymine
- C = Cytosine
- G = Guanine

These bases form the genetic sequence.

For example, we can represent a small DNA sequence as:

A T G C C A T G A

A genetic change can alter the DNA sequence.

For example:

Original sequence:

A T G C C A T G A

Changed sequence:

A T G C T A T G A

Here, one DNA base has changed.

This is a simplified example of a **mutation** or genetic alteration.

However, DNA changes are not limited to changing one base.

Genetic changes can also involve:

- deletion of DNA segments
- copying of DNA segments
- rearrangement of DNA
- changes affecting larger chromosome regions

### Important

A DNA change does NOT automatically mean cancer.

Most genetic changes are not harmful by themselves.

Cancer can develop when important genetic changes accumulate and affect genes involved in processes such as cell growth, cell division, DNA repair, or cell survival.

## 5. Where Can DNA Changes Come From?

Cancer-related genetic changes can arise in different ways.

### 1. Random errors during cell division

Before a cell divides, its DNA must be copied.

The copying process is highly accurate, but occasional errors can occur.
```
Cell
↓
DNA is copied
↓
Cell divides
↓
Occasional copying error
↓
DNA change
```
### 2. Environmental exposures

Certain environmental exposures can damage DNA.

Examples include:

- chemicals in tobacco smoke
- ultraviolet radiation
- certain viruses such as HPV

DNA damage does not automatically become cancer.

Cells have mechanisms that can repair DNA damage or eliminate severely damaged cells.

### 3. Inherited genetic changes

Some genetic changes can be inherited from a parent through an egg or sperm cell.

Therefore, genetic changes can be:

- acquired during a person's lifetime
- inherited from a parent

Most cancer-causing genetic changes are acquired over time rather than directly inherited.

## 6. Which Types of Genes Are Important in Cancer?

Not every DNA change has the same effect.

Some genetic changes affect genes that control important cellular processes.

Three important categories are:

### 1. Proto-oncogenes

Proto-oncogenes normally help cells grow and divide when needed.

If certain changes alter these genes, they can become oncogenes and contribute to excessive cell growth or survival.

Think of them as a **growth accelerator**.

### 2. Tumor suppressor genes

Tumor suppressor genes help control cell growth and division.

They can be thought of as **brakes**.

If these genes are damaged or inactivated, an important growth-control mechanism can be weakened.

### 3. DNA repair genes

DNA repair genes help maintain the integrity of DNA.

If DNA repair mechanisms are impaired, additional genetic changes can accumulate more easily.

Think of them as a **repair system**.

### Simple analogy

Growth genes → accelerator

Tumor suppressor genes → brakes

DNA repair genes → repair system

Cancer can develop when multiple important genetic changes disrupt these control systems.

## 7. Example: BRCA1 and BRCA2

BRCA1 and BRCA2 are genes involved in repairing damaged DNA.

Some people inherit a harmful variant in one of these genes.

This does not mean that the person automatically has breast cancer.

Instead, the inherited variant can increase the risk of developing breast cancer.

A simplified mechanism is:
```
Inherited harmful BRCA1/BRCA2 variant

One copy of the gene is altered
↓
The cell may still have a normal copy
↓
The normal copy can later become altered or lost in a cell
↓
DNA repair ability can become insufficient
↓
Additional genetic changes may accumulate
↓
Cancer can develop
```

There is usually no single factor that allows us to say:

A risk factor is something that increases the probability of developing a disease.

For breast cancer, documented risk factors include:

- certain radiation exposure
- alcohol consumption
- physical inactivity
- excess body weight, particularly after menopause

## **Understanding The dataset**

### Mean feature 

> #### 1. Radius : "How big is it from the center to the edge?"

Radius = distance from the center to the boundary.

If the distance is larger, the tumor is generally larger.

> #### 2. Texture— "How much does the inside of the tumor vary?"

Texture is not the physical roughness of the tumor.
It comes from the image pixels.

Imagine we look at small parts of the tumor image:

Dark  Dark  Gray  Gray  Light

 20    22    50    52     80

If pixel values are very similar: 50  51  50  52  51

→ texture variation is small.

If they vary a lot: 10  30  80  20  90

→ texture variation is larger.

So:

Texture = how much the gray levels/pixel intensities vary across the tumor.

The feature comes from the digital image, not from touching the tumor.

> #### 3.Perimeter:"How long is the boundary around the tumor?"

Perimeter = length of the tumor's outside boundary.

Example
Tumor A                  Tumor B
```
   _____                  _/\__
 /       \              /      \
|         |            |        \
 \_______/              \__/\___/
```
Perimeter ≈ 20          Perimeter ≈ 30

Tumor B has a more complicated boundary, so its perimeter can be larger even if the overall size isn't dramatically different.


> #### 4.Area — "How much space does the tumor occupy?"

Area = the amount of 2D space inside the tumor boundary.

A bigger tumor usually has a bigger radius, perimeter, and area.

> #### 5. Smoothness — "How smoothly does the boundary change?"
Now we look closely at the boundary.

Imagine drawing the tumor boundary with a pencil.

Smooth boundary
```
        ______
      /        \
     /          \
    |            |
     \          /
      \________/
```
The boundary changes gradually.

Irregular boundary
```
       __/\____
     /   \  /   \
    |     \/     |
     \__/\___/\_/
```
The boundary changes more from one point to another.

The border changes more irregularly.
Smoothness measures local variation in the tumor boundary/radius.

- Simple idea:
Is the border changing smoothly, or are there many small irregular changes?

> #### 6. Compactness — "How efficiently is the shape packed?"

Imagine two objects with roughly the same area.

Shape A — compact
```
      ______
    /        \
   |          |
    \________/
```
It's close to a circle.

Shape B — less compact
```
      ______
    /        \
   |          |____
    \_____________/
```
It has a more stretched/irregular boundary.

The compactness feature combines perimeter and area.

In this dataset, the original feature is derived from:
compactness = (perimeter^2/area)-1 

So it asks, roughly:

How complicated/stretched is the boundary compared with the amount of area inside it?

> #### 7. Concavity — "Does the boundary curve inward?"

First imagine a shape with no indentation:
```
       ______
     /        \
    |          |
     \________/
```
The boundary mostly curves outward.

Now imagine this:
```
       ______
     /        \
    |   \      |
    |    \     |
     \_________/
```
There is an indentation going inward.

That inward curve is called concavity.

So:

Concavity measures how much the tumor boundary curves inward

> #### 8. Concave points — "How many important inward points are there?"

This is related to concavity, but not exactly the same.

Imagine:
```

      _________
    /           \
   |    \       |
   |     \__    |
   |        \   |
    \__________/
```
There are points along the boundary where the shape changes into an inward curve.

Those are concave points.

Think of the difference like this:

Concavity :How deep/severe is the inward curve?

Concave points :Where/how many points on the boundary are associated with inward curves?

> #### 9. Symmetry — "Does one side look like the other?"

Imagine cutting a shape down the middle.

Symmetric
```
        /\
       /  \
      /    \
     |      |
      \    /
       \  /
        \/

       │
       │
      cut
```
Left and right sides are approximately similar.

Asymmetric
```
        /\
       /  \
      /    |____
     |          \
      \          |
       \________/
```
Left ≠ Right

So:

Symmetry measures how similar different sides of the tumor are

> #### 10. Fractal dimension — "How complicated is the boundary?"

Simple boundary
```
       _______
     /         \
    |           |
     \_________/
```
Very simple.

Highly irregular boundary
```
     _/\__/\___
   /  \_/  \_/ \
  | /\  /\_/   |
   \_/\/__/\__/
```
The second boundary has much more detail.

So:

Fractal dimension describes the complexity/irregularity of the boundary.

### **Decision Trees do not need feature scaling.**

Unlike distance-based algorithms (KNN, SVM with RBF kernel, K-Means, etc.), a Decision Tree does not calculate distances between points. It simply asks threshold questions on individual features, for example:

Is radius_mean <= 14.2?

Is area_mean <= 650?

Is concavity_mean <= 0.08?

Because each split looks at only one feature at a time, the absolute scale of that feature does not affect the decision. Whether radius_mean is 15 or 0.015, the tree can still find the best threshold.

Therefore:

StandardScaler → not needed

MinMaxScaler → not needed

Any other feature scaling → not needed

## Manual Gini calculation for one feature

````python
def calculate_gini(n_yes,n_no):
    total = n_yes + n_no
    if total == 0:
        return 0.0
    
    p_yes = n_yes / total
    p_no  = n_no  / total
    
    return 1.0 - p_yes**2 - p_no**2

y_array=np.asarray(y)
total_yes=np.sum(y_array==1)
total_no=np.sum(y_array==0)
root_gini=calculate_gini(total_yes,total_no)
print(f"Root Gini:{root_gini:.4f}")
````

````python
def evaluate_all_splits(X, y):
    all_splits = []

    y_array = np.asarray(y) # convert y to a Numpy array for fast indexing 

    for feature in X.columns:
        values = X[feature].to_numpy(dtype=float)

        # Sort values from smallest to largest
        order = np.argsort(values) 
        sorted_values = values[order]
        sorted_y = y_array[order]

        # Candidate boundaries only between distinct values 
        """  
        Example :
        sorted_values = [10, 12, 12, 15, 18, 18, 20]
        
        position bewteen 0 and 1 : 10-12 ,we can put a threshold
        .................1 and 2 : 12-12  ,No (indentical)...
        
        only valide position are [0,2,3,5]
        """
        valid_positions = np.where(sorted_values[:-1] < sorted_values[1:])[0]

        if len(valid_positions) == 0:
            continue

        # Cumulative number of malignant and benign samples
        malignant = (sorted_y == 1).astype(int)
        benign = (sorted_y == 0).astype(int)

        cum_yes = np.cumsum(malignant)
        cum_no = np.cumsum(benign)

        total_yes = malignant.sum()
        total_no = benign.sum()
        total = len(sorted_y)

        for pos in valid_positions:
            # Threshold halfway between adjacent distinct values
            threshold = (sorted_values[pos] + sorted_values[pos + 1]) / 2

            # Left child: feature <= threshold
            left_n = pos + 1
            left_yes = int(cum_yes[pos])
            left_no = int(cum_no[pos])

            # Right child: feature > threshold
            right_n = total - left_n
            right_yes = int(total_yes - left_yes)
            right_no = int(total_no - left_no)

            # Calculate child Gini impurities
            gini_left = calculate_gini(left_yes,left_no)
            gini_right = calculate_gini(right_yes,right_no)

            # Weighted Gini
            weighted_gini = ((left_n / total) * gini_left+ (right_n / total) * gini_right)

            all_splits.append({
                "feature": feature,
                "threshold": threshold,
                "left_yes_M": left_yes,
                "left_no_B": left_no,
                "left_total": left_n,
                "gini_left": gini_left,
                "right_yes_M": right_yes,
                "right_no_B": right_no,
                "right_total": right_n,
                "gini_right": gini_right,
                "weighted_gini": weighted_gini,
                "gini_reduction": root_gini - weighted_gini
            })

    return pd.DataFrame(all_splits)
all_splits = evaluate_all_splits(X, y)

print("Total candidate splits tested:", len(all_splits))
print("Features evaluated:", X.shape[1])
````

````python
# Find the best threshold for each feature
best_per_feature = (
    all_splits
    .sort_values("weighted_gini")
    .groupby("feature", as_index=False)
    .first()
    .sort_values("weighted_gini")
    .reset_index(drop=True)
)

columns_to_show = [
    "feature",
    "threshold",
    "left_yes_M",
    "left_no_B",
    "left_total",
    "gini_left",
    "right_yes_M",
    "right_no_B",
    "right_total",
    "gini_right",
    "weighted_gini",
    "gini_reduction"
]

print(best_per_feature[columns_to_show].round(4).to_string(index=False))
````

````python
features = best_per_feature["feature"].tolist()

fig, axes = plt.subplots(6, 5, figsize=(22, 20))
axes = axes.flatten()

for i, feature in enumerate(features):
    row = best_per_feature.loc[best_per_feature["feature"] == feature].iloc[0]
    ax = axes[i]
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

    # Root node
    ax.text(0.5, 0.82,
            f"{feature}\n≤ {row['threshold']:.4f}",
            ha="center", va="center", fontsize=8,
            bbox=dict(boxstyle="round,pad=0.4", fc="lightblue", ec="black"))

    #  Branches 
    ax.plot([0.5, 0.22], [0.70, 0.52], color="black", lw=1.2)
    ax.plot([0.5, 0.78], [0.70, 0.52], color="black", lw=1.2)

    # Left child
    ax.text(0.22, 0.38,
            f"TRUE (≤)\n"
            f"M: {int(row['left_yes_M'])}\n"
            f"B: {int(row['left_no_B'])}\n"
            f"Gini: {row['gini_left']:.3f}",
            ha="center", va="center", fontsize=7.5,
            bbox=dict(boxstyle="round,pad=0.35", fc="#e8f5e9", ec="black"))

    #Right child
    ax.text(0.78, 0.38,
            f"FALSE (>)\n"
            f"M: {int(row['right_yes_M'])}\n"
            f"B: {int(row['right_no_B'])}\n"
            f"Gini: {row['gini_right']:.3f}",
            ha="center", va="center", fontsize=7.5,
            bbox=dict(boxstyle="round,pad=0.35", fc="#ffebee", ec="black"))

    #Weighted Gini
    ax.text(0.5, 0.08,
            f"Weighted Gini: {row['weighted_gini']:.4f}",
            ha="center", va="center", fontsize=8, fontweight="bold")

# Hide any unused axes (if less than 30 features)
for j in range(len(features), len(axes)):
    axes[j].axis("off")

fig.suptitle("Best Root Split for Each Feature", fontsize=16, y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
````

````python
def predict_diagnosis(model, sample):
    prediction = model.predict(sample)[0]
    probabilities = model.predict_proba(sample)[0]
    if prediction == 0:
        diagnosis = "Benign"
    else:
        diagnosis = "Malignant"

    print("Prediction:", diagnosis)
    print(f"Benign probability: "f"{probabilities[0] * 100:.2f}%")
    print(f"Malignant probability: "f"{probabilities[1] * 100:.2f}%")
````

The ROC curve (Receiver Operating Characteristic) shows how well a model separates the two classes (Malignant vs Benign) at every possible decision threshold

The orange dashed line is the random classifier (AUC = 0.5).

Anything above it is better than random guessing.

The blue line is the Decision Tree.

````python
# train vs test
train_accuracy = []
test_accuracy = []

depths = range(1, 16)

for depth in depths:
    tree = DecisionTreeClassifier(max_depth=depth,random_state=42)
    tree.fit(X_train,y_train)

    train_accuracy.append(tree.score(X_train,y_train))
    test_accuracy.append(tree.score(X_test,y_test))
````

````python
plt.figure(figsize=(10, 6))

plt.plot(depths,train_accuracy,marker="o",label="Training Accuracy")

plt.plot(depths,test_accuracy,marker="o",label="Testing Accuracy")
plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")

plt.title("Decision Tree: Overfitting vs Tree Depth")
plt.legend()
plt.grid(True)

plt.show()
````

- Blue line (Training Accuracy)
Keeps rising and reaches 100 % as the tree gets deeper.

→ The model memorizes the training data.

- Orange line (Testing Accuracy)
Improves until depth ≈ 5–7 (best performance ≈ 100 %)

Then drops and stays around 95 % for deeper trees.

Depth 5–7 → best balance (good on both train and test)
Depth > 7 → overfitting: the tree becomes too complex, fits noise in the training set, and generalizes worse on new data.

The Wisconsin Breast Cancer dataset is relatively easy to separate:

Many features (especially concave points_mean, area_worst, radius_worst…) are highly discriminative.

Even a very shallow tree (depth = 1 or 2) already reaches ~90 % accuracy.

So the “bad” models are still quite good.
